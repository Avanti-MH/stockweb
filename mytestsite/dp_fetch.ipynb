```python
import os
import django
from asgiref.sync import sync_to_async  # 匯入 sync_to_async

# 設定 Django 環境
os.environ.setdefault("DJANGO_SETTINGS_MODULE", "mytestsite.settings")  # 替換為你的專案名稱.settings
django.setup()

from stockapp.models import UserSubscription

async def fetch_subscriptions():
    # 使用 sync_to_async 包裹 ORM 操作
    subscriptions = await sync_to_async(list)(UserSubscription.objects.all())

    # 格式化輸出
    for sub in subscriptions:
        print(f"User: {sub.user.username}")
        print(f"Stock Code: {sub.stock_code}")
        print(f"Parameters: {sub.parameters}")
        print(f"Created At: {sub.created_at}")
        print("-" * 30)

# 在異步環境中執行
import asyncio
if __name__ == "__main__":
    asyncio.run(fetch_subscriptions())
```
